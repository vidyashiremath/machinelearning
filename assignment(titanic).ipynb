```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Load Titanic dataset
df = pd.read_csv("tested.csv")

# 2. Display basic information
print("First 5 rows:")
print(df.head())

print("\nShape of dataset:")
print(df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

# 3. Remove duplicate rows
df = df.drop_duplicates()

print("\nShape after removing duplicates:")
print(df.shape)

# 4. Handle missing values

# Age - replace missing values with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Embarked - replace missing values with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Fare - replace missing values with median
df["Fare"] = df["Fare"].fillna(df["Fare"].median())

# Cabin - create a new feature indicating whether cabin information exists
df["CabinKnown"] = df["Cabin"].notna().astype(int)

# 5. Drop unnecessary columns
df = df.drop(
    columns=["PassengerId", "Name", "Ticket", "Cabin"],
        errors="ignore"
        )

print("\nColumns after dropping unnecessary features:")
print(df.columns)

# 6. Encode categorical variables

# Sex: male = 0, female = 1
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})

# One-hot encode Embarked
df = pd.get_dummies(
            df,
                columns=["Embarked"],
                    drop_first=True,
                        dtype=int
                        )

print("\nDataset after encoding:")
print(df.head())

# 7. Separate input features and target
X = df.drop("Survived", axis=1)
y = df["Survived"]

# 8. Feature scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
                            X_scaled,
                                columns=X.columns
                                )

print("\nScaled features:")
print(X_scaled.head())

# 9. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
                                    X_scaled,
                                        y,
                                            test_size=0.2,
                                                random_state=42,
                                                    stratify=y
                                                    )

# 10. Display final results
print("\nTraining data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nFinal prepared dataset:")
print(X_scaled.head())

print("\nRemaining missing values:")
print(X_scaled.isnull().sum().sum())
```
